## data type convertion

```python
from pyspark.sql.functions import to_date,col
df = spark.table("pl_workforce_catalog.bronze.company") \
    .withColumn("company_id",col("company_id").cast("int")) \
    .withColumn("established_date",to_date("established_date", "dd-mm-yyyy"))

df.write.format("delta").mode("overwrite").saveAsTable("pl_workforce_catalog.silver.company")     
display(df.limit(10))
```
