## data type convertion

```python
from pyspark.sql.functions import col,to_date
df = spark.table("pl_workforce_catalog.bronze.department") \
     .withColumn("department_id", col("department_id").cast("int"))

df.write.format("delta").mode("overwrite").saveAsTable("pl_workforce_catalog.silver.department")          
display(df.limit(15))
```
