# general lergers

## data type convertion

```python
from pyspark.sql.functions import col,substring,to_date

df = spark.table("pl_workforce_catalog.bronze.general_ledgers") \
    .withColumn("entry_date", substring("entry_date", 0, 10)) \
    .withColumn("entry_date", to_date("entry_date", "dd-mm-yyyy")) \
    .withColumn("posting_date",substring("posting_date", 0, 10)) \
    .withColumn("posting_date", to_date("posting_date", "dd-mm-yyyy")) \
    .withColumn("gl_id", col("gl_id").cast("int")) \
    .withColumn("fiscal_year", col("fiscal_year").cast("int")) \
    .withColumn("fiscal_period", col("fiscal_period").cast("int")) \
    .withColumn("company_id", col("company_id").cast("int")) \
    .withColumn("account_id", col("account_id").cast("int")) \
    .withColumn("department_id", col("department_id").cast("int")) \
    .withColumn("debit_amount", col("debit_amount").cast("float")) \
    .withColumn("credit_amount", col("credit_amount").cast("float")) \
    .withColumn("created_by", col("created_by").cast("int")) \
    .withColumn("approved_by", col("approved_by").cast("int"))

display(df.limit(15))
```

## removing inconsistant data

```python
df = df.filter((col("debit_amount") != 0) | (col("credit_amount") != 0)) \
    .filter(col("posting_date") >= col("entry_date"))
df.write.mode("overwrite").saveAsTable("pl_workforce_catalog.silver.general_ledgers")
display(df.limit(15))
```
