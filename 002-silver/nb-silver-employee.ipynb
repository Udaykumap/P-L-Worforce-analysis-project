# employee

## data types convertion

```python
from pyspark.sql.functions import col,substring,to_date
df = spark.table("pl_workforce_catalog.bronze.employee") \
     .withColumn("employee_id", col("employee_id").cast("int")) \
     .withColumn("department_id", col("department_id").cast("int")) \
     .withColumn("company_id", col("company_id").cast("int")) \
     .withColumn("hire_date",substring("hire_date", 0, 10)) \
     .withColumn("hire_date", to_date("hire_date", "dd-MM-yyyy")) \
     .withColumn("base_salary", col("base_salary").cast("float")) 
df.write.mode("overwrite").saveAsTable("pl_workforce_catalog.silver.employee")         
display(df.limit(100))
```
