## data type convertion

```python
from pyspark.sql.functions import col

df = spark.table("pl_workforce_catalog.bronze.accounts") \
    .withColumn("account_id", col("account_id").cast("int")) \
    .dropDuplicates(["account_id"])

df.write.format("delta").mode("overwrite").saveAsTable("pl_workforce_catalog.silver.accounts")
display(df.limit(10))
```
