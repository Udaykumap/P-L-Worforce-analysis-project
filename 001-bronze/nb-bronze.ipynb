## load all the tables

```python
files_tables = {
    "accounts.csv": "pl_workforce_catalog.bronze.accounts",
    "company.csv": "pl_workforce_catalog.bronze.company",
    "departments.csv": "pl_workforce_catalog.bronze.department",
    "employee.csv": "pl_workforce_catalog.bronze.employee",
    "general ledgers.csv": "pl_workforce_catalog.bronze.general_ledgers",
    "payroll.csv": "pl_workforce_catalog.bronze.payroll"
}

for file, table in files_tables.items():
    df = spark.read.format("csv").option("header", "true").load(f"s3://pl-workforce-project-s3-bucket/{file}")
    df.write.format("delta").mode("overwrite").saveAsTable(table)
```
