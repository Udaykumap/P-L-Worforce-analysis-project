# creating dimentional tables

## account

```python
%sql
CREATE OR REPLACE TABLE pl_workforce_catalog.gold.dim_account AS
SELECT
  account_id,
  account_name,
  account_type,
  category
FROM pl_workforce_catalog.silver.accounts;
select * from pl_workforce_catalog.gold.dim_account limit 10;
```

## company 

```python
%sql
CREATE OR REPLACE TABLE pl_workforce_catalog.gold.dim_company AS
SELECT
  company_id,
  company_name,
  industry
FROM pl_workforce_catalog.silver.company;

SELECT * FROM pl_workforce_catalog.gold.dim_company LIMIT 10;
```

## department

```python
%sql
CREATE OR REPLACE TABLE pl_workforce_catalog.gold.dim_department AS
SELECT
  department_id,
  department_name
FROM pl_workforce_catalog.silver.department;

SELECT * FROM pl_workforce_catalog.gold.dim_department LIMIT 10;
```

## employee

```python
%sql
CREATE OR REPLACE TABLE pl_workforce_catalog.gold.dim_employee AS
SELECT
  employee_id,
  first_name,
  last_name,
  department_id,
  company_id,
  position,
  is_active
FROM pl_workforce_catalog.silver.employee;

SELECT * FROM pl_workforce_catalog.gold.dim_employee LIMIT 10;
```
