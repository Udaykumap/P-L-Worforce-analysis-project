# creating fact tables

## general ledgers

```python
%sql
CREATE OR REPLACE TABLE pl_workforce_catalog.gold.fact_general_ledgers AS
SELECT
  gl_id,
  company_id,
  account_id,
  department_id,
  posting_date,
  debit_amount,
  credit_amount,
  created_by,
  approved_by,
  status
FROM pl_workforce_catalog.silver.general_ledgers;

SELECT * FROM pl_workforce_catalog.gold.fact_general_ledgers LIMIT 10;
```

## payroll 

```python
%sql
CREATE OR REPLACE TABLE pl_workforce_catalog.gold.fact_payroll AS
SELECT
  payroll_id,
  employee_id,
  company_id,
  department_id,
  gross_salary,
  bonus,
  overtime_pay,
  commission,
  allowances,
  tax_deduction,
  social_security,
  health_insurance,
  retirement_contribution,
  other_deductions,
  net_salary,
  status,
  pay_date
FROM pl_workforce_catalog.silver.payroll;

SELECT * FROM pl_workforce_catalog.gold.fact_payroll LIMIT 10;
```
