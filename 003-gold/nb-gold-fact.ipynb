# creating fact tables

## general ledgers

```python
%sql
CREATE OR REPLACE TABLE pl_workforce_catalog.gold.fact_general_ledgers AS
SELECT
  gl_id,
  company_id,
  account_id,
  department_id,
  posting_date,
  debit_amount,
  credit_amount,
  created_by,
  approved_by,
  status
FROM pl_workforce_catalog.silver.general_ledgers;

SELECT * FROM pl_workforce_catalog.gold.fact_general_ledgers LIMIT 10;
```

## payroll 

```python
%sql
CREATE OR REPLACE TABLE pl_workforce_catalog.gold.fact_payroll AS
SELECT
  p.payroll_id,
  p.employee_id,
  p.company_id,
  p.department_id,
  p.gross_salary,
  p.bonus,
  p.overtime_pay,
  p.commission,
  p.allowances,
  p.tax_deduction,
  p.social_security,
  p.health_insurance,
  p.retirement_contribution,
  p.other_deductions,
  p.net_salary,
  p.status,
  p.pay_date,
  e.base_salary
FROM pl_workforce_catalog.silver.payroll p
LEFT JOIN pl_workforce_catalog.silver.employee e
  ON p.employee_id = e.employee_id;

SELECT * FROM pl_workforce_catalog.gold.fact_payroll;
```
