```python
%sql
CREATE OR REPLACE TABLE pl_workforce_catalog.gold.master_cube AS
SELECT
    c.company_id,
    c.company_name,
    d.department_id,
    d.department_name,
    e.employee_id,
    e.position,
    e.is_active,
    a.account_id,
    a.account_type,
    a.category,
    f.posting_date,
    f.credit_amount,
    f.debit_amount,
    p.pay_date,
    p.base_salary,
    p.gross_salary,
    p.bonus,
    p.overtime_pay,
    p.commission
FROM pl_workforce_catalog.gold.dim_employee e
JOIN pl_workforce_catalog.gold.dim_company c ON e.company_id = c.company_id
JOIN pl_workforce_catalog.gold.dim_department d ON e.department_id = d.department_id
LEFT JOIN pl_workforce_catalog.gold.fact_payroll p ON e.employee_id = p.employee_id
LEFT JOIN pl_workforce_catalog.gold.fact_general_ledgers f ON e.company_id = f.company_id
    AND YEAR(p.pay_date) = YEAR(f.posting_date)
    AND MONTH(p.pay_date) = MONTH(f.posting_date)
LEFT JOIN pl_workforce_catalog.gold.dim_account a ON f.account_id = a.account_id;

SELECT * FROM pl_workforce_catalog.gold.master_cube;
```
